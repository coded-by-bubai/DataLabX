```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,15,30]
plt.plot(x,y)
plt.show()
```

```python
x=['Mon', 'tue', 'wed', 'thu', 'fri']
y = [10,15,7,20,12]
plt.plot(x,y)
plt.title("Bakery sales this week")
plt.xlabel("day")
plt.ylabel("sales")

plt.show()
```

### plotting line 

```python
# plt.plot(x,y, color= "color name", linestyle = "style", linewidth = "width", marker= "symbol", label = "label name")
m = [1,2,3,4]
sales = [1000,1800,1600,2000]
plt.plot(m, sales, color="blue", linestyle=":", linewidth = 2, marker="o", label = "2025 sales data")

plt.title("sales report")
plt.xlabel("month")
plt.ylabel("sales")

plt.legend(loc="lower right", fontsize=7)
plt.grid(color="gray", linestyle="--", linewidth=1)
plt.ylim(500, 2500)
plt.xticks(m,['M1','M2','M3','M4'])

plt.show()
```

### plotting bar, pie, hist

```python
# plt.bar(x,h,color="color name", width = "value", label ="name")
product  = ['A','B', 'C', 'D']
sales = [1000,1500,800,1200]

plt.barh(product, sales, color = "orange", label = "sales ")

plt.title("sales data")
plt.xlabel("product")
plt.ylabel("product sales")
plt.legend()
plt.show()
```

```python
# plt.pie(value, label_list, color=color list, autopct='%1.1f%%')

region = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1000 ,1500]

plt.pie(revenue, labels=region, autopct='%1.1f%%')
plt.title("revenue contribution by region")
plt.show()
```

```python
# plt.hist(data, bins=number of bins, color=colorname, edgecolor = 'black')
score = [45,67,89,85,78,95,45,60,65,32,58,75,55]
plt.hist(score, bins=5, color='blue', edgecolor='yellow')
plt.xlabel("score")
plt.ylabel("frequency")
plt.title("score distribution")
plt.show()
```

### scatter plot

```python
# plt.scatter(x,y,color = colorname, marker = 'symbol', label=name)
hours = [1,2,3,5,6,8,9,10,15]
exam_score = [50,55,45,85,65,89,57,87,90]

plt.scatter(hours, exam_score)
plt.xlabel("hours")
plt.ylabel("score")
plt.grid()
plt.show()
```

```python
# plt.scatter(x,y,color = colorname, marker = 'symbol', label=name)

plt.scatter([1,2,3], [89,85,90], color='orange', label = "g1")
plt.scatter([1,2,3], [55,75,60], color='blue', label="g2")
plt.xlabel("hours")
plt.ylabel("score")
plt.title("comparison")
plt.legend()
plt.grid()
plt.show()
```

### subplot

```python
# plt.Subplot(row, column, index)
x = [1,2,3,4]
y=[10,35,30,40]
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("line")
plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("bar")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1,2,figsize=(10,5))
x = [1,2,3,4]
y=[10,35,30,40]
ax[0].plot(x,y)
ax[0].set_title("line")
ax[1].bar(x,y)
ax[1].set_title("bar")

fig.suptitle("comparison")
plt.tight_layout()
plt.show()
```
