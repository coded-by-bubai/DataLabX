5. IPL Player Runs Analysis

| Player | Runs |
| ------ | ---- |
| A      | 550  |
| B      | 670  |
| C      | 490  |
| D      | 720  |
| E      | 610  |


Task

1. Create horizontal bar chart.
2. Sort players by runs.
3. Highlight top scorer.

```python
df = pd.DataFrame({
    "Player": ["A", "B", "C", "D", "E"],
    "Runs": [550, 670, 490, 720, 610]
})
df.sort_values("Runs", ascending=False, inplace=True)
colors = ["gold"] + ["C0"]*df.count()["Player"]
bars = plt.barh(df["Player"], df["Runs"], color = colors)
plt.bar_label(bars)
plt.ylabel("Player")
plt.xlabel("Runs")
plt.show()
```

6. Company Profit Analysis

| Year | Profit |
| ---- | ------ |
| 2019 | 25     |
| 2020 | 30     |
| 2021 | 45     |
| 2022 | 50     |
| 2023 | 65     |

Task

1. Plot profit growth.
2. Calculate growth percentage.
3. Annotate highest profit point.

```python
df = pd.DataFrame({
    "Year": [2019, 2020, 2021, 2022, 2023],
    "Profit": [25, 30, 45, 50, 65]
})
max_idx = df["Profit"].idxmax()
max_year = df.loc[max_idx, "Year"]
max_profit = df.loc[max_idx, "Profit"]
df["Growth"] = df["Profit"].pct_change()*100
plt.plot(df["Year"], df["Profit"], marker="o")
plt.annotate(f"Highest: {max_idx}",
             xy=(max_year, max_profit ),
             xytext=(max_year , max_profit + 5),
            arrowprops=dict(arrowstyle="->")
             )
plt.show()
```

7. Age Distribution Histogram

Generate 500 random ages between 18 and 60.

Task

1. Create histogram.
2. Use different bin sizes.
3. Identify most common age range.

```python
import numpy as np
Ages = np.random.randint(18,61,500)
counts, bins, patches = plt.hist(Ages, bins=range(18, 61, 8), width=5)
max_idx = counts.argmax()
patches[max_idx].set_facecolor('orange')
print(f"Most common age range: {bins[max_idx]:.0f}-{bins[max_idx+1]:.0f}")
plt.annotate(
    f"most common\n{bins[max_idx]:.0f}-{bins[max_idx+1]:.0f}",
    xy=((bins[max_idx]+bins[max_idx+1]) / 2 - 2, counts[max_idx]),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)
plt.show()
```

8. Height vs Weight Scatter Plot

Generate:

Height (150-190 cm)<br>
Weight (45-100 kg)

for 200 people.

Task

1. Create scatter plot.
2. Find correlation.
3. Add trend line.

```python
height = np.random.randint(150, 190, 200)
weight = np.random.randint(45, 100, 200)
df= pd.DataFrame({
    "Height": height,
    "Weight": weight
})
corr = df["Height"].corr(df["Weight"])
print(f"Correlation: {corr:.3f}")
plt.scatter(height, weight)

m, b = np.polyfit(df["Height"], df["Weight"], 1)
plt.plot(df["Height"], m*df["Weight"]+b)
plt.xlabel("height")
plt.ylabel("weight")
```

10. COVID-19 Data Visualization

Use a real COVID dataset.

Task

1. Daily cases line chart.
2. Monthly trend chart.
3. Rolling average plot.
4. Compare two countries.

```python
# Load dataset
df = pd.read_csv("day_wise.csv")
df["Date"] = pd.to_datetime(df["Date"])
print(df.info())
plt.plot(df["Date"], df["New cases"])
plt.show()
```

```python
monthly = df.groupby(df["Date"].dt.to_period('M'))["New cases"].sum().reset_index()
print(monthly)
plt.plot(monthly["Date"].dt.strftime("%Y-%m"), monthly["New cases"], marker = "o")
# plt.ylim(5000)
plt.ticklabel_format(style='plain', axis='y')
# plt.show()
```

```python
df["rolling_avg"] = df["New cases"].rolling(window=7).mean()
print(df[["Date", "New cases", "rolling_avg"]].head(10))
plt.plot(df["Date"],df["rolling_avg"],label="rolling 7 day avg")
plt.plot(df["Date"],df["New cases"], label="daily cases")
plt.legend()
plt.show()
```
