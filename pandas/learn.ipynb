# handling missing data

```python
# NAN
# None
# isnull() return true -> nan is exit, false nan is not exit 
data = {
    "Name" : ["a",'b', None, 'c'],
    "age": [20,21, None, 23],
    "salary": [5000,6000, None, 3000],
    "performane" : [45,98, None, 58]
}
df = pd.DataFrame(data)
print(df)
print(df.isnull())
print(df.isnull().sum())
```

```python
# .dropna(axis = 1, inplace= True)
df.dropna(inplace=True)
print(df)
```

```python
# .fillna(value, inplace = True)
print(df)
df["salary"].fillna(df["salary"].mean(), inplace=True)
print(df)
df.fillna(0,inplace=True)
print(df)
```

```python
# interpolation - linear, polynomial
# 10, 20, nan(30 fill), 40, 50
data = {
    "time" : [1,2,3,4,5,6],
    "value" : [10, None, 30, None, 50, None]
}
df = pd.DataFrame(data)
print(df)
# .interpolate(method="linear",axis=0, inplace=True )
df["value"] = df["value"].interpolate(method="linear" )
print(df)
```

# sorting

```python
# .sort_values(by="cname", ascending = true/false, inplace =true)
data = {
    "Name" : ["a",'b', None, 'c','d'],
    "age": [25,21, None, 23,23],
    "salary": [5000,6000, None, 3000,4000],
    "performane" : [45,98, None, 58,98]
}
df = pd.DataFrame(data)
print(df)

df.sort_values(by="age",ascending=True, inplace=True)
print(df)
df.sort_values(by=["age", "salary"],ascending=[True,False], inplace=True)
print(df)
```

# group

```python
data = {
    "Name" : ["a",'b', 'c','d','e'],
    "age": [25,21, 23,23,21],
    "salary": [5000,6000, 3000,4000,5000],
    "performane" : [45,98, 58,98, 58]
}
df = pd.DataFrame(data)
print(df)
grouped = df.groupby(["age", "Name"])["salary"].sum()
print(grouped)
grouped = df.groupby("age")["salary"].sum()
print(grouped)
```

# merging

```python
# pd.merge(df1, df2,on="column", how="type of join")
df_customer = pd.DataFrame({
    'CustomerID' : [1,2,3],
    'Name':["ramesh", "suresh", "kalpesh"]
})
df_order = pd.DataFrame({
    'CustomerID' : [1,2,4],
    'orderAmount':[250,300,500]
})
# merge
df_merged = pd.merge(df_customer,df_order, how="inner", on="CustomerID")
print("inner join: \n",df_merged)
df_merged = pd.merge(df_customer,df_order, how="outer", on="CustomerID")
print("outer join: \n",df_merged)
df_merged = pd.merge(df_customer,df_order, how="left", on="CustomerID")
print("left join: \n",df_merged)
df_merged = pd.merge(df_customer,df_order, how="right", on="CustomerID")
print("right join: \n",df_merged)
df_merged = pd.merge(df_customer,df_order, how= "cross")
print("cross join: \n",df_merged)
```

# concatenate

```python
# pd.concat([df1,df2],axis=0, ignore_index=True)
region1 = pd.DataFrame({
    'CustomerID' : [1,2,3],
    'Name':["ramesh", "suresh", "kalpesh"]
})
region2 = pd.DataFrame({
    'CustomerID' : [4,5,6],
    'Name':["h", "s", "k"]
})
df_concate = pd.concat([region1, region2], axis=0)
print(df_concate)
df_concate = pd.concat([region1, region2], axis=0, ignore_index=True)
print(df_concate)
```
