4. Missing Values

Given:
```python
data = {
    'Name':['A','B','C','D'],
    'Marks':[80,None,75,None]
}
```
Tasks:

1. Find missing values.
2. Count missing values.
3. Replace missing values with average marks.

`````python
data = {
    'Name':['A','B','C','D'],
    'Marks':[80,None,75,None]
}
df = pd.DataFrame(data)
# Find missing values.
print("find:\n",df.isnull())
# Count missing values.
print("Count: \n",df.isnull().sum())
# Replace missing values with average marks.
df.fillna(df['Marks'].mean(), inplace=True)
print(df)
`````

5. Sorting Data

|D     |  50000|

Tasks:

1. Sort salary ascending.

6. Student Performance Analysis

Create a DataFrame:

|Name	|Math	|Science	|English|
|-------|-------|-----------|-------|
|A	|85	|78	|92
|B	|70	|88	|80
|C	|95	|90	|85
|D	|60	|75	|70

Tasks:

1. Calculate total marks.
2. Calculate average marks.
3. Find topper.
4. Find students scoring above 80 average.

`````python
df = pd.DataFrame({
    "Name":['A','B','C', 'D'],
    "Math":[85,70,95,60],
    "Science":[78,88,90,75],
    "English":[92,80,85,70]
})
# Calculate total marks.
df["Total"] = df[["Math","Science", "English"]].sum(axis=1)
print(df[["Name", "Total"]])
# Calculate average marks.
df["Avg"] = df[["Math","Science", "English"]].mean(axis=1)
print(df)
# Find topper.
print("Topper: ", df["Name"][df["Total"] == df["Total"].max()])
# Find students scoring above 80 average.
print("Student avg > 80:\n",df["Name"][df["Avg"] > 80])
`````

8. GroupBy Practice

Dataset:

|Product	|Category	|Sales|
|-----------|-----------|-----|
|Laptop	|Electronics	|50000|

|Shirt	|Clothing	|5000|
|Jeans	|Clothing	|7000|

Tasks:

1. Group by category.
2. Find total sales per category.
3. Find average sales per category.

`````python
df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Shirt", "Jeans"],
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing"],
    "Sales": [50000, 30000, 5000, 7000]
})

print(df)

# 1. Group by category.
g_cat = df.groupby(by="Category", as_index=False)
# 2. Find total sales per category.
print("total : \n",g_cat["Sales"].sum())
# 3. Find average sales per category.
print("avg: \n", g_cat["Sales"].mean())
`````

11. Sales Data Analysis

Dataset Columns:
```python
Date
Product
Region
Sales
```
Tasks:

1. Monthly sales report.
2. Region-wise sales.
3. Best-selling product.
4. Highest revenue month.

`````python
df = pd.DataFrame({
    "Date": ["2025-02-01", "2025-01-02", "2025-02-03", "2025-01-04", "2025-01-05"],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone"],
    "Region": ["North", "South", "East", "West", "North"],
    "Sales": [50000, 30000, 45000, 20000, 35000]
})
df["Date"] = pd.to_datetime(df["Date"])
print(df)
print("Monthly sales report.")
monthly_sales = df.groupby(df["Date"].dt.to_period('M'))["Sales"].sum().reset_index()
print(monthly_sales)
print("Region-wise sales.")
regeon_sales = df.groupby("Region")["Sales"].sum().reset_index()
print(regeon_sales)
print("Best-selling product.")
best_sale_pro = df.groupby("Product")["Sales"].sum().reset_index().sort_values(by="Sales", ascending=False).head(1)
print(best_sale_pro)
print("Highest revenue month.")
high_rev_m = monthly_sales.sort_values(by="Sales", ascending=False).head(1)["Date"].dt.strftime("%B")
print(high_rev_m)
`````

13. E-Commerce Dataset

Columns:
```python
OrderID
CustomerID
Category
Price
Quantity
```
Tasks:

1. Create Revenue column.

Formula:

> Revenue = Price * Quantity
2. Find category-wise revenue.
3. Top-selling category.
4. Top 10 highest-value orders.

`````python
df = pd.DataFrame({
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006],
    "CustomerID": [101, 102, 103, 101, 104, 102],
    "Category": [
        "Electronics",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Electronics"
    ],
    "Price": [50000, 1500, 30000, 12000, 2500, 45000],
    "Quantity": [1, 2, 1, 1, 3, 1]
})

print(df)
# Create Revenue column.
df["Revenue"] = df["Price"] * df["Quantity"]
print(df)
print("Find category-wise revenue.")
cat_rev = df.groupby("Category")["Revenue"].sum().reset_index()
print(cat_rev)
print("Top-selling category: ", end="")
print(cat_rev.sort_values(by="Revenue", ascending=False).head(1)["Category"].to_string(index=False))
# Top 10 highest-value orders.
print(df.nlargest(10, "Revenue").sort_values("Revenue", ascending=False))
`````

14. Time Series Analysis

Dataset:
```
Date
Temperature
```
Tasks:

1. Convert Date to datetime.
2. Find average monthly temperature.
3. Find hottest day.
4. Calculate 7-day rolling average.

`````python
df = pd.DataFrame({
    "Date": [
        "2025-01-01", "2025-01-03", "2025-01-05",
        "2025-01-08", "2025-01-10", "2025-01-15",
        "2025-01-20", "2025-01-25", "2025-02-01",
        "2025-02-05", "2025-02-10", "2025-02-15",
        "2025-02-20", "2025-02-25", "2025-03-01",
        "2025-03-05", "2025-03-10", "2025-03-15"
    ],
    "Temperature": [
        22, 24, 23,
        35, 26, 24,
        23, 22, 27,
        28, 29, 30,
        31, 29, 32,
        33, 34, 35
    ]
})
# print(df)
# Convert Date to datetime.
df["Date"] = pd.to_datetime(df["Date"])
# Find average monthly temperature.
print("monthly avg: \n", df.groupby(df["Date"].dt.to_period("M"))["Temperature"].mean().reset_index())
# Find hottest day.
print("Hot day :\n",df[df["Temperature"] == df["Temperature"].max()])
# Calculate 7-day rolling average.
print("Weekly avg: \n", df.groupby(df["Date"].dt.to_period("W"))["Temperature"].mean().reset_index())
`````
